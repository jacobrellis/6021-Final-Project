# Data Cleaning

## Inspection

```python
import pandas as pd
import numpy as np
```

```python
sleep.dtypes
```

```python
sleep['sleep_quality_score'].isna().sum()
```

sleep_quality_score target has nulls. Need to drop those observations. Other potential targets don't have nulls.

## Drop redundant/necessary columns

```python
# cant keep nulls on a target
sleep = sleep.dropna(subset=['sleep_quality_score'])
sleep = sleep.drop(columns='respondent_id')
len(sleep)
```

```python
sleep.head()
```

## Map yes/no to 0 and 1

```python
sleep['doomscroller'].unique()
```

```python
# verified that these only contain yes/no
binary_cols = ['uses_night_mode', 'keeps_phone_in_bedroom', 'consumes_negative_news_content', 'uses_sleep_tracking_app', 'doomscroller']
binary_map = {'Yes': 1, 'No': 0}


sleep[binary_cols] = sleep[binary_cols].replace(binary_map)
```

## Wrangle categoricals

```python
categorical_nulls = ['occupation_status', 'primary_device_used_at_night', 'bedtime_routine_type']

sleep[categorical_nulls] = sleep[categorical_nulls].fillna('Unknown')

sleep.head()
```

```python
sleep['sleep_quality_category'] = sleep['sleep_quality_category'].astype('category')

sleep.dtypes
```

## Wrangle numerics: MEAN IMPUTATION

```python
numerics_to_update = ['caffeine_intake_mg_per_day', 'exercise_minutes_per_day']

sleep[numerics_to_update] = sleep[numerics_to_update].fillna(sleep[numerics_to_update].mean())
```
