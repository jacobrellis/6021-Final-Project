```python
import pandas as pd
import re
from pathlib import Path
```

```python

sas = Path("data/2025XXH-SAS-Input-Program.sas").read_text(
    encoding="utf-8-sig"
)

# Extract only the SAS INPUT section
input_block = re.search(r"\nInput\n(.*?);", sas, re.I | re.S).group(1)

# Extract start position, name, character flag, width, and decimals
fields = re.findall(
    r"@(\d+)\s+(\w+)\s+(\$?)(\d+)\.(\d*)",
    input_block,
)

fields
```

```python
df = pd.read_fwf(
    "data/XXH2025_YRBS_Data.dat",
    colspecs=[
        (int(start) - 1, int(start) - 1 + int(width))
        for start, name, char, width, decimals in fields
    ],
    names=[name for start, name, char, width, decimals in fields],
    header=None,
    dtype="string",
    keep_default_na=False,
    na_values=["", "."],
)

# Convert SAS numeric fields, including implied decimal places
for start, name, char, width, decimals in fields:
    if char != "$":
        raw = df[name]
        values = pd.to_numeric(raw, errors="raise")

        if decimals and int(decimals) > 0:
            has_decimal = raw.str.contains(".", regex=False, na=False)
            values = values.where(has_decimal, values / 10 ** int(decimals))

        df[name] = values

```
